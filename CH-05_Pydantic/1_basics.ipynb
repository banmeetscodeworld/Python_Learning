```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    id:int
    name:str
    department:str
```

```python
#it is a pydantic object, obj will return Pydantic object/pydantic schema
obj = employee_schema(id=1, name="John Doe", department="IT")
obj
```

```python
#passing arguments as a dictionary
data = {"id": 1, "name": "John Doe", "department": "IT"}
obj = employee_schema(**data)
obj
```

```python
#this is data validation by Pydantic, it will raise an error if the data type is incorrect
#in this case, department should be a string, but an integer is provided, so it will raise a validation error
obj2 = employee_schema(id=1, name="John Doe", department= 101)
obj2
```

## **Coercion VS Strict Types**

## Coercion

```python
#Lineant behaviour
#Pydantic tries to convert the data to the correct type if possible
obj3 = employee_schema(id="101", name="John Doe", department="IT")
obj3
```

## Strict Types

```python
#now this will not simply convert the string "101" to an integer because we are using StrictInt and will throw an error
from pydantic import BaseModel, StrictInt

class employee_schema(BaseModel):
    id: StrictInt
    name:str
    department:str

obj4 = employee_schema(id = "101",name="John Doe", department="HR")
obj4
```
